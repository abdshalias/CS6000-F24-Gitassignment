```python
import random
import numpy as np
import matplotlib.pyplot as plt

# Generate random numbers
def generate_random_numbers(n=1000, lower=0, upper=100):
    """Generates 'n' random numbers between 'lower' and 'upper'."""
    random_numbers = [random.randint(lower, upper) for _ in range(n)]
    return random_numbers

# Calculate basic statistics
def calculate_statistics(numbers):
    """Calculates mean and standard deviation of the generated numbers."""
    mean = np.mean(numbers)
    std_dev = np.std(numbers)
    return mean, std_dev

# Plot the random numbers
def plot_random_numbers(numbers):
    """Plots a histogram of the generated random numbers."""
    plt.hist(numbers, bins=30, alpha=0.7, color='blue', edgecolor='black')
    plt.title('Histogram of Generated Random Numbers')
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.show()

# Main function
if __name__ == "__main__":
    n = 1000  # Number of random numbers
    random_numbers = generate_random_numbers(n, lower=0, upper=100)

    # Calculate and print statistics
    mean, std_dev = calculate_statistics(random_numbers)
    print(f"Generated {n} random numbers.")
    print(f"Mean: {mean}")
    print(f"Standard Deviation: {std_dev}")

    # Plot the generated random numbers
    plot_random_numbers(random_numbers)
```
